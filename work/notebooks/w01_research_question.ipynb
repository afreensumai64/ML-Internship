## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset provides page-level observations that can be used to investigate whether there is enough variation in search visibility and performance to support a review-prioritization problem. I will use the dataset to measure the size of the available sample and identify pages that receive search impressions but no clicks, which is one observable pattern that may warrant further investigation.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the starter dataset directly from your GitHub repository
url = "https://raw.githubusercontent.com/afreensumai64/ML-Internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

# 1. Pages with 100+ impressions but 0 clicks
zero_click = df[
    (df["impressions_90d"] >= 100) &
    (df["clicks_90d"] == 0)
]

print("Pages with 100+ impressions but 0 clicks:", len(zero_click))

# 2. Of those, how many are on page 1?
page_1_zero_click = zero_click[
    zero_click["position_tier"] == "page_1"
]

print(
    "Page-1 pages with 100+ impressions and 0 clicks:",
    len(page_1_zero_click)
)

# 3. Percentage
print(
    "Percentage of zero-click pages that are on page 1:",
    round(len(page_1_zero_click) / len(zero_click) * 100, 2),
    "%"
)
```
